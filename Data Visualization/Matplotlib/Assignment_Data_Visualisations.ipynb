```python
# Problem-1
import numpy as np
import matplotlib.pyplot as plt

scores = np.random.normal(loc=70, scale=10, size=200)

plt.figure()
plt.hist(scores, bins=20, color="green")
plt.axvline(scores.mean(), linestyle='--', color="red")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.title("Histogram of Scores with Mean Line")
plt.show()
```

```python
## Probelm-2

import seaborn as sns
import matplotlib.pyplot as plt

penguins =sns.load_dataset("penguins")

plt.figure()
sns.scatterplot(
    penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species"
)

plt.title("Flipper Length vs Body Mass(Penguins)")
plt.show()
```

```python
## problem 3

tips = sns.load_dataset("tips")

plt.figure()
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.title("Total Bill by Day and Sex")
plt.show()
```

```python
## Problem-4

plt.figure()
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker"
)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
fig, ax=plt.subplots()

sns.scatterplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    ax=ax
)

ax.set_title("avg bill value for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")

fig.tight_layout()
```
