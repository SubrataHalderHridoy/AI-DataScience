## LR with OLS(Ordinary LastSquares)

```python
class LinearRegressionOLS:
    def __init__(self):
        self.bias = None
        self.weights = None

    def fit(self, X, y):
        m, n = X.shape

        X_updated = np.c_[np.ones((m, 1)), X]

        # normal equation
        # theta = np.dot(np.linealg.inv(np.dot(X_updated.T, X_upadted)), np.dot(X_updated.T, y))
        theta = np.linalg.inv(X_updated.T @ X_updated) @ X_updated.T @ y

        self.bias = theta[0]
        self.weights = theta[1:]

    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegressionOLS()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weights)
```
